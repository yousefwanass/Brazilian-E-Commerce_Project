```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
sales = pd.read_csv("D:/Data Analysis/Tamweely Train/Brazilian E-Commerce Dataset/data/sales_master.csv")

sales.head()
```

```python
sales.info()
```

```python
sales.isnull().sum()
```

```python
print("Duplicates Before :", sales.duplicated().sum())

sales = sales.drop_duplicates()

print("Duplicates After :", sales.duplicated().sum())
```

```python
sales["payment_value"] = sales["payment_value"].fillna(
    sales["payment_value"].median()
)
```

```python
sales["order_purchase_timestamp"] = pd.to_datetime(
    sales["order_purchase_timestamp"]
)

sales["order_delivered_customer_date"] = pd.to_datetime(
    sales["order_delivered_customer_date"]
)
sales["order_delivered_carrier_date"] = pd.to_datetime(
    sales["order_delivered_carrier_date"]
)
sales["order_estimated_delivery_date"] = pd.to_datetime(
    sales["order_estimated_delivery_date"]
)
sales["shipping_limit_date"] = pd.to_datetime(
    sales["shipping_limit_date"]
)
```

```python
sales.info()
```

```python
sales["Delivery_Days"] = (
    sales["order_delivered_customer_date"]
    - sales["order_purchase_timestamp"]
).dt.days
```

```python
sales["Order_Month"] = sales["order_purchase_timestamp"].dt.month
```

```python
sales["Order_Year"] = sales["order_purchase_timestamp"].dt.year
```

```python
sales = sales.rename(columns={

    "customer_city":"City",

    "customer_state":"State",

    "payment_value":"Payment",

    "price":"Revenue"

})
```

```python
sales["Order_Year"] = sales["Order_Year"].astype("int")

sales["Order_Month"] = sales["Order_Month"].astype("int")

sales["Delivery_Days"] = sales["Delivery_Days"].astype("float")
```

```python
sales["product_category_name"] = (

    sales["product_category_name"]

    .str.lower()

    .str.strip()

)
```

```python
Q1 = sales["Payment"].quantile(.25)

Q3 = sales["Payment"].quantile(.75)

IQR = Q3-Q1

lower = Q1-1.5*IQR

upper = Q3+1.5*IQR

outliers = sales[
    (sales["Payment"]<lower) |
    (sales["Payment"]>upper)
]

print(outliers.shape)
```

```python
sales = sales[
    (sales["Payment"]>0) &
    (sales["Revenue"]>0)
]
```

```python
sales[["order_id", "order_item_id"]].head(20)
```

```python
sales["order_item_id"].value_counts().sort_index()
```

```python
sales["quantity"] = (
    sales.groupby("order_id")["order_item_id"]
         .transform("max")
)
```

```python
sales[["order_id", "order_item_id", "quantity"]].head(20)
```

```python
sales[sales["order_item_id"] > 1][["order_id", "order_item_id"]].head(20)
```

```python
sales.groupby("order_id")["order_item_id"].max().sort_values(ascending=False).head(20)
```

```python
sales["quantity"] = (
    sales.groupby("order_id")["order_item_id"]
    .transform("max")
)
```

```python
sales.loc[
    sales["order_id"] == "8272b63d03f5f79c56e9e4120aec44ef",
    ["order_id", "order_item_id", "quantity"]
]
```

```python
sales["Total_Revenue"] = sales["Payment"] * sales["quantity"]
```

```python
sales["Profit"] = sales["Total_Revenue"] * 0.20
```

```python
sales["Delivery_Status"] = np.where(

    sales["Delivery_Days"]<=7,

    "Fast",

    "Delayed"

)
```

```python
sales["Customer_Type"] = np.where(

    sales.duplicated("customer_id"),

    "Returning",

    "New"

)
```

```python
report = {

"Missing Values":sales.isnull().sum(),

"Duplicate Count":sales.duplicated().sum(),

"Outliers":len(outliers),

"Data Types":sales.dtypes,

"Cleaned Records":len(sales)

}

report
```

```python
sales.head()
```

```python
sales["Delivery_Days"] = sales["Delivery_Days"].fillna(
    sales["Delivery_Days"].median()
)
```

```python
sales["product_category_name"] = sales["product_category_name"].fillna("Unknown")
```

```python
sales["product_weight_g"] = sales["product_weight_g"].fillna(
    sales["product_weight_g"].median()
)
```

```python
sales["product_length_cm"] = sales["product_length_cm"].fillna(
    sales["product_length_cm"].median()
)
```

```python
sales["product_height_cm"] = sales["product_height_cm"].fillna(
    sales["product_height_cm"].median()
)
```

```python
sales["product_width_cm"] = sales["product_width_cm"].fillna(
    sales["product_width_cm"].median()
)
```

```python
sales.drop(columns=["product_name_lenght"], inplace=True)
```

```python
sales.drop(columns=["product_description_lenght"], inplace=True)
```

```python
sales.drop(columns=["product_photos_qty"], inplace=True)
```

```python
sales.drop(
    columns=[
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date"
    ],
    inplace=True
)
```
