```python
customers = pd.read_csv(r"D:\Data Analysis\Tamweely Train\Brazilian E-Commerce Dataset\data\olist_customers_dataset.csv")
orders = pd.read_csv(r"D:\Data Analysis\Tamweely Train\Brazilian E-Commerce Dataset\data\olist_orders_dataset.csv")
order_items = pd.read_csv(r"D:\Data Analysis\Tamweely Train\Brazilian E-Commerce Dataset\data\olist_order_items_dataset.csv")
payments = pd.read_csv(r"D:\Data Analysis\Tamweely Train\Brazilian E-Commerce Dataset\data\olist_order_payments_dataset.csv")
products = pd.read_csv(r"D:\Data Analysis\Tamweely Train\Brazilian E-Commerce Dataset\data\olist_products_dataset.csv")
```

```python
datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Payments": payments,
    "Products": products
}

for name, df in datasets.items():
    print("="*60)
    print(name)
    print(df.head())
```

```python
for name, df in datasets.items():

    print("="*70)
    print(name)

    print("\nRows:", df.shape[0])
    print("Columns:", df.shape[1])

    print("\nData Types")
    print(df.dtypes)

    print("\nMissing Values")
    print(df.isnull().sum())

    print("\nUnique Values")
    print(df.nunique())

    print("\nDuplicate Rows")
    print(df.duplicated().sum())
```

```python
sales_master = customers.merge(
    orders,
    on="customer_id",
    how="inner"
)

sales_master = sales_master.merge(
    order_items,
    on="order_id",
    how="inner"
)

sales_master = sales_master.merge(
    payments,
    on="order_id",
    how="inner"
)

sales_master = sales_master.merge(
    products,
    on="product_id",
    how="left"
)
```
